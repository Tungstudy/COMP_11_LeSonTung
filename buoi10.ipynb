```python
#1
def merge(arr, vec):
  res = []
  id = 0
  i = 0
  while True:
    if id == len(arr) or i == len(arr):
      res += arr[id - 1:]
      res += vec[i - 1:]
      return res
    if arr[id] <= vec[i]:
      res.append(arr[id])
      id += 1
    else:
      res.append(vec[i])
      i += 1
arr = list(map(int, input().split()))
vec = list(map(int , input().split()))
print(merge(arr, vec))



```

```python
#3
def swap(i,j,num):
    if i > j:
        i,j = j,i
    new = num[:i] + num[j] + num[i+1:j] + num[i] + num[j+1:]
    return new

def solve(arr):
    ans = 10**30
    for i in range(len(arr) - 1):
        for j in range(i, len(arr)):
            new_int = int(swap(i,j,arr))
            if new_int < ans:
                ans = new_int
    return ans
```

```python
#6
def check(n,target):
  return n > target

def compare(arr, target):
  # trả về index
  lo = 0
  hi = len(arr) - 1
  ans = 0
  while lo <= hi:
    mid = (lo + hi)//2
    if check(arr[mid], target):
      ans = mid
      lo = mid + 1
    else:
      hi = mid - 1
  return ans

def lo_xo(lst,n):
  is_greater = False
  id = 0
  for i in range(n - 1,0, -1):
    if lst[i] > lst[i - 1]:
      is_greater = True
      id = i - 1
      break
  if is_greater:
    id_smallest = compare(lst[id +1:], lst[id]) + id + 1
     # đoạn này có thể thay thế bằng tìm kiếm nhị phân nhưng lười + đã thay!!!
     #có thể duyệt ngược mảng!
    # change
    lst[id_smallest], lst[id] = lst[id], lst[id_smallest]
    new = lst[:id + 1] + lst[id +1:][::-1]
    return new
  else:
    return lst[::-1]
n = int(input())

lst = list(map(int,input().split()))
print(lo_xo(lst,n))
```

```python
#7
def can_be_planted(arr, k):
  ans = 0
  if arr[0] == 0:
    ans += 1

  for i in range(1, len(arr) - 1):
    if arr[i] == 0 and arr[i + 1] == 0:
      arr[i] = 1
      ans += 1
  if arr[-1] == 0:
    ans += 1
  return ans >= k
arr = list(map(int,input().split()))
k = int(input())

print(can_be_planted(arr,k))
```

```python
#8
class List:
  def __init__(self,x,y):
    self.x = x
    self.y = y
    self.vector = (self.x, self.y)
  def __add__(self, other):
    self.x += other.x
    self.y += other.y
    self.vector = (self.x, self.y)
    return self
point = {(0,0): 1}
right = List(1,0)
left = List(-1,0)
up = List(0,1)
down = List(0,-1)
#start
start = List(0,0)
def is_repeat(arr):
  global right, left, up, down,start
  for item in arr:
    if item == 'R':
      start += right
      if not start.vector in point:
        point[start.vector] = 1
      else:
        return True
    elif item == 'L':
      start += left
      if not start.vector in point:
        point[start.vector] = 1
      else:
        return True
    elif item == 'U':
      start += up
      if not start.vector in point:
        point[start.vector] = 1
      else:
        return True
    elif item == 'D':
      start += down
      if not start.vector in point:
        point[start.vector] = 1
      else:
        return True
  return False

arr = input()
print(is_repeat(arr))
```
