```python
#3
def sum_all_digit(num):
    str_num = str(num)
    sum = 0
    for item in str_num:
        sum += int(item)**2
    return sum
def is_happy(num):
    try:
        data = {}
        while num != 1:
            num = sum_all_digit(num)
            if num not in data:
                data[num] = 1
            else:
                raise ValueError
        return True
    except ValueError:
        return False
print(is_happy(11))

```

```python
#5
def change_to_int(lst):
    for i in range(len(lst)):
        try:
            lst[i] = int(lst[i])
        except (TypeError, ValueError):
            print(f"Error for index: {i}")
    return lst
lst = ['1',2,3,4,5,'a','!']
print(change_to_int(lst))
```

```python
#7
def age():
    try:
        n = int(input())
        if n < 0:
            raise ValueError
        return 2025 - n
    except ValueError:
        print("Invalid Error")

age()
```
